Anscome's Quartet

```python
import pandas as pd
#import matplotlib.pyplot as plt
import numpy as np
import plotnine as pn
```

```python
# Dataset I
df1 = pd.DataFrame({
    'x': [10.0, 8.0, 13.0, 9.0, 11.0, 14.0, 6.0, 4.0, 12.0, 7.0, 5.0],
    'y': [8.04, 6.95, 7.58, 8.81, 8.33, 9.96, 7.24, 4.26, 10.84, 4.82, 5.68]
})

# Dataset II
df2 = pd.DataFrame({
    'x': [10.0, 8.0, 13.0, 9.0, 11.0, 14.0, 6.0, 4.0, 12.0, 7.0, 5.0],
    'y': [9.14, 8.14, 8.74, 8.77, 9.26, 8.10, 6.13, 3.10, 9.13, 7.26, 4.74]
})

# Dataset III
df3 = pd.DataFrame({
    'x': [10.0, 8.0, 13.0, 9.0, 11.0, 14.0, 6.0, 4.0, 12.0, 7.0, 5.0],
    'y': [7.46, 6.77, 12.74, 7.11, 7.81, 8.84, 6.08, 5.39, 8.15, 6.42, 5.73]
})

# Dataset IV
df4 = pd.DataFrame({
    'x': [8.0, 8.0, 8.0, 8.0, 8.0, 8.0, 8.0, 19.0, 8.0, 8.0, 8.0],
    'y': [6.58, 5.76, 7.71, 8.84, 8.47, 7.04, 5.25, 12.50, 5.56, 7.91, 6.89]
})
```

```python
# List of dataframes and titles
datasets = [df1, df2, df3, df4]
titles = ['I', 'II', 'III', 'IV']

# Set same axis limits for all plots
x_min, x_max = 0, 20
y_min, y_max = 0, 14
```

```python
# Combine datasets into one table for faceting
df_all = pd.concat(
    [df.assign(dataset=title) for df, title in zip(datasets, titles)],
    ignore_index=True,
    axis=0,
    )

# Plot: scatterplots for all four datasets
(
    pn.ggplot(df_all, pn.aes(x='x', y='y'))
    + pn.geom_point(size=2)
    + pn.facet_wrap('~dataset', nrow=2)
    + pn.coord_cartesian(xlim=(x_min, x_max), ylim=(y_min, y_max))
    + pn.labs(title="Anscombe's Quartet", x='x', y='y')
    + pn.theme_bw()
    + pn.theme(figure_size=(9, 6))
)
```
