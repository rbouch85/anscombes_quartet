# Anscombe's Quartet Visualization

Creating four charts of Anscombe's Quartet with minimalist styling.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
```

## Data from Wikipedia

Data points from: https://en.wikipedia.org/wiki/Anscombe%27s_quartet

```python
# Dataset I
df1 = pd.DataFrame({
    'x': [10.0, 8.0, 13.0, 9.0, 11.0, 14.0, 6.0, 4.0, 12.0, 7.0, 5.0],
    'y': [8.04, 6.95, 7.58, 8.81, 8.33, 9.96, 7.24, 4.26, 10.84, 4.82, 5.68]
})

# Dataset II
df2 = pd.DataFrame({
    'x': [10.0, 8.0, 13.0, 9.0, 11.0, 14.0, 6.0, 4.0, 12.0, 7.0, 5.0],
    'y': [9.14, 8.14, 8.74, 8.77, 9.26, 8.10, 6.13, 3.10, 9.13, 7.26, 4.74]
})

# Dataset III
df3 = pd.DataFrame({
    'x': [10.0, 8.0, 13.0, 9.0, 11.0, 14.0, 6.0, 4.0, 12.0, 7.0, 5.0],
    'y': [7.46, 6.77, 12.74, 7.11, 7.81, 8.84, 6.08, 5.39, 8.15, 6.42, 5.73]
})

# Dataset IV
df4 = pd.DataFrame({
    'x': [8.0, 8.0, 8.0, 8.0, 8.0, 8.0, 8.0, 19.0, 8.0, 8.0, 8.0],
    'y': [6.58, 5.76, 7.71, 8.84, 8.47, 7.04, 5.25, 12.50, 5.56, 7.91, 6.89]
})
```

## Create Visualizations

Creating four separate charts with minimalist styling.

```python
# List of dataframes and titles
datasets = [df1, df2, df3, df4]
titles = ['I', 'II', 'III', 'IV']

# Set same axis limits for all plots
x_min, x_max = 0, 20
y_min, y_max = 0, 14

# Create a separate figure for each dataset
for idx, (df, title) in enumerate(zip(datasets, titles)):
    # Create a new figure for this dataset with 10x8 layout
    fig, ax = plt.subplots(figsize=(10, 8))
    
    # Calculate linear regression
    slope, intercept, r_value, p_value, std_err = stats.linregress(df['x'], df['y'])
    
    # Create regression line
    x_line = np.array([x_min, x_max])
    y_line = slope * x_line + intercept
    
    # Plot regression line (light grey)
    ax.plot(x_line, y_line, color='lightgrey', linewidth=1.5, zorder=1)
    
    # Plot data points (blue) - twice as large
    ax.scatter(df['x'], df['y'], color='blue', s=100, zorder=2)
    
    # Set axis limits
    ax.set_xlim(x_min, x_max)
    ax.set_ylim(y_min, y_max)
    
    # Add tick marks for whole-integer values
    ax.set_xticks(range(x_min, x_max + 1))
    ax.set_yticks(range(y_min, y_max + 1))
    
    # Remove tick labels (minimalist approach)
    ax.set_xticklabels([])
    ax.set_yticklabels([])
    
    # Add title
    ax.set_title(title, fontsize=12)
    
    # Keep the spines for minimal frame
    for spine in ax.spines.values():
        spine.set_edgecolor('black')
        spine.set_linewidth(0.5)
    
    # Adjust layout
    plt.tight_layout()
    
    # Save the figure with dataset-specific filename
    filename = f'anscombes_quartet_{title}.png'
    plt.savefig(filename, dpi=300, bbox_inches='tight')
    print(f"Chart saved as '{filename}'")
    
    # Display the figure
    plt.show()
    
    # Close the figure to free memory
    plt.close(fig)
```
